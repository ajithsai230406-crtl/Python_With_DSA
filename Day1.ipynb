```python
#9. My solution
'''Given an array and q range-sum queries, return the sum for each query.

For query (l, r), calculate:

arr[l] + arr[l+1] + ... + arr[r]

Use 0-based indexing.'''
def range_sum_queries(arr, queries):
    prefix_sum = [0] * (len(arr) + 1)
    for i in range(len(arr)):
        prefix_sum[i + 1] = prefix_sum[i] + arr[i]
    result = []
    for l, r in queries:
        result.append(prefix_sum[r + 1] - prefix_sum[l])
    return result
print("Range sum queries result:", range_sum_queries([1, 2, 3, 4, 5], [(0, 2), (1, 3), (2, 4)]))    
```

```python
#10. My solution
'''Given a sorted array and a target value, determine whether there are two distinct elements whose sum equals the target.'''
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return False
print("Two distinct elements sum to target:", two_sum_sorted([1, 2, 3, 4, 5], 9))
```

```python
#15. My solution
'''Given two strings, determine whether they are anagrams of each other.'''
def are_anagrams(str1,str2): #using DSA
    str1 = str1.replace(" ","").lower()
    str2 = str2.replace(" ","").lower()
    if len(str1) != len(str2):
        return False
    freq = {}
    for char1 in str1:
        freq[char1] = freq.get(char1,0) + 1
    for char2 in str2:
        if char2 not in freq:
            return False
        freq[char2] -= 1
    return True and all(count == 0 for count in freq.values())
print("Are the two strings Anagrams to each other?:",are_anagrams("Dormitory","Dirty room"))
```

```python
#17. My solution
'''Given an array of positive integers and a target S, find the length of the shortest contiguous subarray whose sum is at least S.'''
def shortest_subarr_with_sum(arr,S):
    n = len(arr)
    min_len  = float('inf')
    current_sum = 0
    left = 0

    for right in range(n):
        current_sum += arr[right]
        while current_sum >= S:
            min_len = min(min_len,right - left + 1)
            current_sum -= arr[left]
            left += 1
    return min_len if min_len != float('inf') else 0
print("Shortest lenght:",shortest_subarr_with_sum([2,3,1,2,4,3],12))
     
```

```python
#17. GPT solution
def shortest_subarray_simple(arr, S):
    shortest = len(arr) + 1  # Start with an impossible maximum length
    current_sum = 0
    left = 0
    
    for right in range(len(arr)):
        # 1. Expand the right side
        current_sum += arr[right]
        
        # 2. Shrink from the left side while the sum is still enough
        while current_sum >= S:
            current_length = right - left + 1
            if current_length < shortest:
                shortest = current_length
                
            current_sum -= arr[left]
            left += 1
            
    # If shortest was never changed, no valid subarray was found
    return shortest if shortest <= len(arr) else 0

print("Shortest length:", shortest_subarray_simple([2, 3, 1, 5, 6, 7],14))
```

```python
#18. My solution
'''You are given a list of log lines:
logs = [
    "ERROR: database connection failed",
    "INFO: user logged in",
    "ERROR: timeout occurred",
    "WARNING: disk usage high",
    "INFO: user logged out",
    "ERROR: database connection failed"
]

Write a function that returns a dictionary containing:
1. Number of INFO, WARNING, and ERROR messages.
2. The most frequently occurring complete log message.'''
from collections import Counter

def analyze_logs(logs):
    counts = {"INFO": 0, "WARNING": 0, "ERROR": 0}
    
    # Count each log level
    for log in logs:
        level = log.split(":", 1)[0].strip()
        if level in counts:
            counts[level] += 1
            
    # Find the most frequent complete log message using Counter
    log_counts = Counter(logs)
    most_frequent = log_counts.most_common(1)[0][0] if logs else None
    
    return {
        "counts": counts,
        "most_frequent": most_frequent
    }

logs = [
    "ERROR: database connection failed",
    "INFO: user logged in",
    "ERROR: timeout occurred",
    "WARNING: disk usage high",
    "INFO: user logged out",
    "ERROR: database connection failed"
]

print(analyze_logs(logs))
```

```python
#19. My solution 
'''Given an integer array, find the length of the longest consecutive sequence of integers.

The numbers do not need to be adjacent in the original array.'''
def longest_consecutive(nums):
    if not nums:
        return 0

    num_set = set(nums)
    longest = 0

    for num in num_set:
        # Only start counting from a sequence's first number
        if num - 1 not in num_set:
            current = num
            streak = 1

            while current + 1 in num_set:
                current += 1
                streak += 1

            longest = max(longest, streak)

    return longest
print("The longest consecutive sequence of intergers is:",longest_consecutive([100, 4, 200, 1, 3, 2]))
```

```python
#19. GPT solution
def longest_consecutive_sorting(nums):
    if not nums:
        return 0
        
    nums.sort()
    longest = 1
    current_streak = 1
    
    for i in range(1, len(nums)):
        # Skip duplicates
        if nums[i] == nums[i - 1]:
            continue
        # If it continues the sequence
        elif nums[i] == nums[i - 1] + 1:
            current_streak += 1
        else:
            # Sequence broken, reset streak
            longest = max(longest, current_streak)
            current_streak = 1
            
    return max(longest, current_streak)

print("The longest consecutive sequence of intergers is:",longest_consecutive_sorting([100, 4, 200, 1, 3, 2]))
```

```python
#21. My solution
'''You are given an array of integers. Find the maximum product of any contiguous subarray.''' 

def max_product_subarray(arr):
    if not arr:
        return None
    max_ending = min_ending = best = arr[0]
    for num in arr[1:]:
        if num < 0:
            max_ending, min_ending = min_ending, max_ending
        max_ending = max(num, max_ending * num)
        min_ending = min(num, min_ending * num)
        best = max(best, max_ending)
    return best
print("Maximum product:", max_product_subarray([2, 3, -2, 4]))
```
