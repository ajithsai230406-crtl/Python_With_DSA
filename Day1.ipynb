```python
#1.My solution
'''Given a list of integers, return the second largest distinct element.'''
def sec_lar_num(num):
    first = second = float('-inf')
    for i in num:
        if i > first:
            second = first
            first = i
        elif first > i > second:
            second = i
    return None if second == float("-inf") else second
arr = [34,52,53,32,43,79]
print(sec_lar_num(arr))
```

```python
#2. My solution
'''Given a string, determine whether it contains only unique characters.'''
def unq_str(str):
    freq = {}
    for char in str:
        if char in freq:
            return False
        freq[char] = 1
    return True
str1 = input("Enter a string :")
if unq_str(str1):
    print(f"The string {str1} is unique")
else:
    print(f"The string {str1} is not unique")
```

```python
#6. My solution
'''Given an array, :move all zeroes to the end while maintaining the relative order of the non-zero elements.'''
def move_zeroes(arr):
    j = 0
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[i],arr[j] = arr[j],arr[i]
            j += 1
    return arr
arr = [0,1,0,13,34,12,43]
print(move_zeroes(arr))
```

```python
#6. GPT solution
'''Given an array, move all zeroes to the end while maintaining the relative order of the non-zero elements.'''
def move_zeros(arr):
    non_zero_idx = 0
    for i in range(len(arr)):
        if arr[i] != 0:
            arr[non_zero_idx] = arr[i]
            non_zero_idx += 1
    for i in range(non_zero_idx,len(arr)):
        arr[i] = 0
    return arr
arr = [1,3,0,34,0,42]
print(move_zeros(arr)) 
```

```python
#7. My solution 
'''Given an integer array, find all elements that appear more than once.'''
def repeated_elements_count(arr):
    freq = {}
    for i in arr:
        freq[i] = freq.get(i,0) + 1
    return {k: v for k,v in freq.items() if v > 1}
print("Repeated elements with their counts:", repeated_elements_count([1, 3, 5, 2, 3, 4, 5, 2, 3]))
```

```python
#8. my solution
'''Given an array of integers, find the maximum sum of any contiguous subarray.'''
def max_subarray_sum(arr):
    max_sum = float('-inf')
    current_sum = 0
    for num in arr:
        current_sum += num
        if current_sum > max_sum:
            max_sum = current_sum
        if current_sum < 0:
            current_sum = 0
    return max_sum
print("Maximum sum of contiguous subarray is:", max_subarray_sum([-2,1,-3,4,-1,2,1,-5,4]))
```

```python
#9. My solution
'''Given an array and q range-sum queries, return the sum for each query.

For query (l, r), calculate:

arr[l] + arr[l+1] + ... + arr[r]

Use 0-based indexing.'''
def range_sum_queries(arr, queries):
    prefix_sum = [0] * (len(arr) + 1)
    for i in range(len(arr)):
        prefix_sum[i + 1] = prefix_sum[i] + arr[i]
    result = []
    for l, r in queries:
        result.append(prefix_sum[r + 1] - prefix_sum[l])
    return result
print("Range sum queries result:", range_sum_queries([1, 2, 3, 4, 5], [(0, 2), (1, 3), (2, 4)]))    
```

```python
#10. My solution
'''Given a sorted array and a target value, determine whether there are two distinct elements whose sum equals the target.'''
def two_sum_sorted(arr, target):
    left, right = 0, len(arr) - 1
    while left < right:
        current_sum = arr[left] + arr[right]
        if current_sum == target:
            return True
        elif current_sum < target:
            left += 1
        else:
            right -= 1
    return False
print("Two distinct elements sum to target:", two_sum_sorted([1, 2, 3, 4, 5], 9))
```

```python
#11. My solution
'''An array contains integers from 1 to n, but one number is missing. Find the missing number.'''
def find_missing_number(arr, n):
    expected_sum = n * (n + 1) // 2
    actual_sum = sum(arr)
    return expected_sum - actual_sum
print("Missing number:", find_missing_number([1, 2, 4, 5,6], 6))
```

```python
#11. GPT solution
def find_missing_number_xor(arr, n):
    xor_sum = 0
    # XOR all numbers from 1 to n
    for i in range(1, n + 1):
        xor_sum ^= i
    # XOR all numbers present in the array
    for num in arr:
        xor_sum ^= num
    return xor_sum

print("Missing number (XOR):", find_missing_number_xor([1, 2, 4, 5, 6], 6))
```

```python
#12. My solution
'''Given an array, find the length of the longest strictly increasing contiguous subarray.'''
def longest_increasing_subarray(arr):
    if not arr:
        return 0
    max_length = 1
    current_length = 1
    for i in range(1, len(arr)):
        if arr[i] > arr[i - 1]:
            current_length += 1
            max_length = max(max_length, current_length)
        else:
            current_length = 1
    return max_length

print("Length of longest strictly increasing contiguous subarray:", longest_increasing_subarray([1, 2, 3, 2, 4, 5, 6]))
```

```python
#12. GPT solution
def longest_increasing_subarray_elements(arr):
    if not arr:
        return []
    
    max_len = 1
    curr_len = 1
    end_idx = 0
    
    for i in range(1, len(arr)):
        if arr[i] > arr[i - 1]:
            curr_len += 1
            if curr_len > max_len:
                max_len = curr_len
                end_idx = i
        else:
            curr_len = 1
            
    start_idx = end_idx - max_len + 1
    return arr[start_idx:end_idx + 1]

print("Longest increasing subarray:", longest_increasing_subarray_elements([1, 2, 3, 2, 4, 5, 6]))
```

```python
#13. My solution
'''Given an array and a target k, count how many pairs of distinct indices (i, j) satisfy: arr[i] + arr[j] == k'''
def count_pairs_with_sum(arr, k):
    num_count = {}
    count = 0
    for num in arr:
        complement = k - num
        if complement in num_count:
            count += num_count[complement]
        num_count[num] = num_count.get(num, 0) + 1
    return count

print("Number of pairs with sum equal to target:", count_pairs_with_sum([1, 2, 3, 4, 5], 5))
```

```python
#13. GPT solution
def count_pairs_brute(arr, k):
    count = 0
    n = len(arr)
    for i in range(n):
        for j in range(i + 1, n):
            if arr[i] + arr[j] == k:
                count += 1
    return count
print("Number of pairs with sum equal to target:", count_pairs_brute([1, 2, 3, 4, 5], 5))
```
