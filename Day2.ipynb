```python
#1. My solution
'''Given a string, find the first character that appears more than once. Return None if every character is unique.'''
def repeated_first_char(str):
    freq = {}
    for char in str:
        freq[char] = freq.get(char,0) + 1
    result = [char for char,count in freq.items() if count > 1]
    return result[0] if result else None
str = "ABCDEGBEC"
print(f"The first character of String '{str}' that appears more than once : {repeated_first_char(str)} ")
```

```python
#1. GPT solution
def repeated_first_char(S):
    seen = set()
    for char in S:
       if char in seen:
          return char
       seen.add(char)
    return None
Str1 = "programming"
print(f"The first character of String '{Str1}' that appears more than once : {repeated_first_char(Str1)} ")
```

```python
 #2. My solution
'''Given a string, find the first character that appears exactly once. Return None if no such character exists.'''
def first_char_exac_once(Str2):
    freq = {}
    for char1 in Str2:
        freq[char1] = freq.get(char1,0) + 1
    for char2 in Str2:
        if freq[char2] == 1:
            return char2
    return None
Str2 = "Leetcode"
print(f"The first character of String '{Str2}' that appears exactly once is : {first_char_exac_once(Str2)}")
                                                 
```

```python
#2. GPT solution
from collections import Counter
def first_char_exac_once(str):
    freq = Counter(str)
    for char in str:
        if freq[char] == 1:
            return char
    return None
Str = "aabbasbdsd"
print(f"The first character of String '{Str}' that appears exactly once is : {first_char_exac_once(Str)}")
```

```python
#3. My solution
'''Given an integer array, find the first element that appears at least twice. Return None if there are no duplicates.'''
def repeated_first_ele_arr(arr):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num,0) + 1
    first_freq = [nums for nums,count in freq.items() if count > 1]
    return first_freq[0] if first_freq else None
arr1 = [1,5,3,4,3,5,6,5,3,2,1]
print(f"The first element of array {arr1} that appears at least twice is : {repeated_first_ele_arr(arr1)} ")
```

```python
#3.GPT solution
def repeated_first_ele_arr(arr):
    seen = set()  #"first repeating element" = first element whose duplicate is encountered during the left-to-right scan.
    for num in arr:
        if num in seen:
            return num
        seen.add(num)
    return None
arr1 = [1,5,3,4,3,5,6,5,3,2,1]
print(f"The first element of array {arr1} that appears at least twice is : {repeated_first_ele_arr(arr1)} ")
```

```python
#4. My solution
'''Given an integer array and a target k, find the number of unique pairs of values whose sum equals k.'''
def unq_pair_values_sum_k(arr, k):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num, 0) + 1

    pairs = set()
    for num in freq:
        complement = k - num
        if complement in freq and num <= complement:
            if num != complement or freq[num] >= 2:
                pairs.add((num, complement))

    return len(pairs)
```
