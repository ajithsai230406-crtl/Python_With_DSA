```python
#1. My solution
'''Given a string, find the first character that appears more than once. Return None if every character is unique.'''
def repeated_first_char(str):
    freq = {}
    for char in str:
        freq[char] = freq.get(char,0) + 1
    result = [char for char,count in freq.items() if count > 1]
    return result[0] if result else None
str = "ABCDEGBEC"
print(f"The first character of String '{str}' that appears more than once : {repeated_first_char(str)} ")
```

```python
#4. My solution
'''Given an integer array and a target k, find the number of unique pairs of values whose sum equals k.'''
def unq_pair_values_sum_k(arr, k):
    freq = {}
    for num in arr:
        freq[num] = freq.get(num, 0) + 1
    pairs = set()
    for num in freq:
        complement = k - num
        if complement in freq and num <= complement:
            if num != complement or freq[num] >= 2:
                pairs.add((num, complement))
    return len(pairs), pairs
arr,k = [1,2,3,4,5],5
count,pairs = unq_pair_values_sum_k(arr, k)
print(f"In array : {arr} number of unique pair values {pairs} whose sum is {k} are of length: {count}")
```

```python
#4.GPT solution
def unq_pair_values_sum_k(arr, k):
    seen = set()
    unique_pairs = set()
    
    for num in arr:
        complement = k - num
        if complement in seen:
            # Sort the pair so (a, b) and (b, a) are treated as identical
            unique_pairs.add(tuple(sorted((num, complement))))
        seen.add(num)
        
    return len(unique_pairs), unique_pairs

arr, k = [1, 2, 3, 4, 5], 6
count,pairs = unq_pair_values_sum_k(arr, k)
print(f"In array : {arr} number of unique pair values {pairs} whose sum is {k} are of length: {count}")
```

```python
#5. My solution
'''Given a string, determine whether it is a palindrome, ignoring spaces, punctuation, and capitalization.'''
def is_palindrome(str):
    left = 0
    right = len(str) - 1
    while left < right:
        while left < right and not str[left].isalnum():
            left += 1
        while left < right and not str[right].isalnum():
            right -= 1
        if str[left].lower() != str[right].lower():
            return False
        left += 1
        right -= 1
    return True
print(is_palindrome("Racecar"))
print(is_palindrome("A man, a plan, a canal: Panama"))  # True
print(is_palindrome("hello"))
```

```python
#5. GPT solution
def is_palindrome(s):
    cleaned = "".join(char.lower() for char in s if char.isalnum())
    return cleaned == cleaned[::-1]
print(is_palindrome("MADam"))                     
print(is_palindrome("A man, a plan, a canal: Panama")) 
print(is_palindrome("NainA"))                       
```

```python
#6. My solution
'''Given a sorted array, remove all duplicates in-place and return the new length.'''
def remove_duplicates(arr):
    if not arr:
        return 0
    arr.sort()
    write = 1
    for read in range(1, len(arr)):
        if arr[read] != arr[write - 1]:
            arr[write] = arr[read]
            write += 1
    return write,arr[:write]

arr = [2,3,4,6,2,4,5,7,2,8,4,3,6]
length, unq = remove_duplicates(arr)
print(f"The new array after removing duplicates {unq} is of lenght: {length}")
```

```python
#6. GPT solution
def remove_duplicates_pythonic(arr):
    # Convert to set to remove duplicates, sort it, and update arr in-place
    arr[:] = sorted(set(arr))
    return len(arr), arr

arr = [2, 3, 4, 6, 2, 4, 5, 7, 2, 8, 4, 3, 6]
length, unq = remove_duplicates_pythonic(arr)

print(f"The new array after removing duplicates {unq} is of length: {length}")
```

```python
#7. My solution
'''Given an integer array, find the length of the longest contiguous subarray whose sum is exactly k.'''
def max_sub_array_len(nums, k):
    prefix_sum_map = {}
    max_len = 0
    current_sum = 0
    
    for i in range(len(nums)):
        current_sum += nums[i]
        
        if current_sum == k:
            max_len = i + 1
            
        if (current_sum - k) in prefix_sum_map:
            length = i - prefix_sum_map[current_sum - k]
            max_len = max(max_len, length)
            
        if current_sum not in prefix_sum_map:
            prefix_sum_map[current_sum] = i
            
    return max_len

nums = [1, -1, 5, -2, 3]
k = 3
print(f"Length of the longest subarray with sum {k}: {max_sub_array_len(nums, k)}")
```

```python
#8. My solution
'''Given an array of positive integers and a target k, find the maximum length of a contiguous subarray whose sum is at most k.'''
def max_sub_array_len_at_most_k(nums, k):
    left = 0
    current_sum = 0
    max_len = 0
    
    for right in range(len(nums)):
        current_sum += nums[right]
        
        while current_sum > k and left <= right:
            current_sum -= nums[left]
            left += 1
            
        max_len = max(max_len, right - left + 1)
          
    return max_len

nums = [1, 2, 3, 4, 1]
k = 7
print(f"Max length of subarray with sum at most {k}: {max_sub_array_len_at_most_k(nums, k)}")
```

```python
#9. My solution
'''Given a string s, find the length of the longest substring without repeating characters.'''
def len_longest_substr(str):
    n = len(str)
    left = 0
    max_len = 0
    char_index = {}

    for right in range(n):
        if str[right] in char_index and char_index[str[right]] >= left:
            left = char_index[str[right]] + 1
        char_index[str[right]] = right
        max_len = max(max_len, right - left + 1)

    return max_len

Str = "abcbdbabdabcdaa" 
print(f"In given String '{Str}' the Longest length of substring without repeating characters: {len_longest_substr(Str)}")

```

```python
#9. GPT solution
def len_longest_substr_set(s):
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
            
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
        
    return max_len

Str = "abcbdbabdabcdaa"
print(f"Max length (Set approach): {len_longest_substr_set(Str)}")
```

```python
#10. My solution
'''Given an integer array nums and an integer k, return the total number of contiguous subarrays whose sum equals k.'''
def subarray_sum(nums, k):
    prefix_counts = {0: 1}
    current_sum = 0
    count = 0
    
    for num in nums:
        current_sum += num
        target = current_sum - k
        if target in prefix_counts:
            count += prefix_counts[target]
        prefix_counts[current_sum] = prefix_counts.get(current_sum, 0) + 1
        
    return count

nums = [1, -1, 1, 1, 1]
k = 2
print(f"Total subarrays with sum {k}: {subarray_sum(nums, k)}")
```

```python
#11. My solution
'''Given an unsorted integer array nums, return the length of the longest sequence of consecutive integers.'''
def longest_consecutive(nums):
    num_set = set(nums)
    max_len = 0
    
    for num in num_set:
        if (num - 1) not in num_set:
            current_num = num
            current_len = 1
            while (current_num + 1) in num_set:
                current_num += 1
                current_len += 1
            max_len = max(max_len, current_len)
    return max_len

nums = [100, 4, 200, 1, 3, 2]
print(f"Longest consecutive sequence length: {longest_consecutive(nums)}")  # Output: 4 ([1, 2, 3, 4])
```

```python
#11. GPT solution
def longest_consecutive_sorting(nums):
    if not nums:
        return 0
    
    nums.sort()
    max_len = 1
    current_len = 1
    for i in range(1, len(nums)):
        if nums[i] == nums[i - 1]:
            continue
        elif nums[i] == nums[i - 1] + 1:
            current_len += 1
        else:
            max_len = max(max_len, current_len)
            current_len = 1
    return max(max_len, current_len)

nums = [100, 4, 200, 1, 3, 2]
print(f"Longest consecutive sequence length: {longest_consecutive_sorting(nums)}")
```

```python
#12. My solution
'''Given an integer array nums and an integer k, return the number of index pairs ((i,j)) such that:
- (i < j)
- nums[i] + nums[j] == k'''
def count_pairs_with_sum(nums, k):
    freq = {}
    count = 0
    
    for num in nums:
        complement = k - num
        if complement in freq:
            count += freq[complement]
        freq[num] = freq.get(num, 0) + 1
    return count

nums = [1, 5, 7, -1, 5]
k = 6
print(f"Number of pairs summing to {k}: {count_pairs_with_sum(nums, k)}")
```

```python
#12. GPT solution
def count_pairs_brute_force(nums, k):
    count = 0
    n = len(nums)
    
    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == k:
                count += 1
                
    return count

nums = [1, 5, 7, -1, 5]
k = 6
print(f"Number of pairs summing to {k}: {count_pairs_brute_force(nums, k)}")
```
